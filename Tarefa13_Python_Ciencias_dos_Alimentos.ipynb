### Distribuição Normal (Gaussiana)

A distribuição Normal é caracterizada por sua forma simétrica e pelo formato de sino, sendo definida por dois parâmetros fundamentais: a média ($\mu$) e o desvio padrão ($\sigma$).

#### Fórmula da Função de Densidade de Probabilidade (PDF)

$$ f(x | μ, σ) = \frac{1}{σ \sqrt{2\pi}} e^{ -\frac{1}{2} \left( \frac{x - μ}{σ} \right)^2 } $$

Onde:
* $x$ é o valor da variável aleatória.
* $μ$ (mi) é a média da distribuição.
* $σ$ (sigma) é o desvio padrão da distribuição.

#### Significado dos Parâmetros $μ$ e $σ$

1. **$μ$ (Média):**
   * Representa o centro da distribuição. Na distribuição Normal, $μ$ corresponde ao centro da distribuição e também à posição do ponto de maior densidade da curva (o topo do sino).
   * **Efeito na Curva:** A alteração de $μ$ desloca a curva horizontalmente ao longo do eixo $X$, sem alterar sua dispersão ou formato.

2. **$σ$ (Desvio Padrão):**
   * Mede a dispersão ou variabilidade dos dados em relação à média.
   * **Efeito na Curva:** Determina a largura e a altura da curva. Um $σ$ maior resulta em uma curva mais larga e achatada (maior variabilidade), enquanto um $σ$ menor resulta em uma curva mais estreita e pontiaguda (dados mais concentrados).

### Entendendo as Funções da `scipy.stats.norm`

* **`pdf(x, loc=mu, scale=sigma)` (Probability Density Function)**:
  * Fornece a densidade de probabilidade em cada ponto. **Sua altura, isoladamente, não representa uma probabilidade**. As probabilidades são obtidas pela área sob a curva em determinado intervalo.
* **`cdf(x, loc=mu, scale=sigma)` (Cumulative Distribution Function)**:
  * Calcula a probabilidade acumulada de a variável assumir um valor menor ou igual a $x$: $P(X \le x)$.
* **`sf(x, loc=mu, scale=sigma)` (Survival Function)**:
  * Calcula a probabilidade de a variável assumir um valor estritamente maior que $x$: $P(X > x) = 1 - P(X \le x)$.
* **`ppf(q, loc=mu, scale=sigma)` (Percent Point Function)**:
  * É a inversa da CDF. Retorna o valor $x$ correspondente a um quantil ou percentil desejado $q$ (por exemplo, qual valor de álcool limita os 90% menores valores).

```python
mu_alcohol = df_white['alcohol'].mean()
sigma_alcohol = df_white['alcohol'].std()

print(f"Média do teor alcoólico dos vinhos brancos (μ): {mu_alcohol:.2f}%")
print(f"Desvio padrão do teor alcoólico dos vinhos brancos (σ): {sigma_alcohol:.2f}%")
```

```python
# Cálculo exato das probabilidades teóricas
mu = mu_alcohol
sigma = sigma_alcohol

# a. Probabilidade acumulada (P(X <= 10))
prob_less_than_10 = norm.cdf(10, loc=mu, scale=sigma)

# b. Probabilidade em um intervalo (P(10 < X <= 12))
prob_between_10_and_12 = norm.cdf(12, loc=mu, scale=sigma) - norm.cdf(10, loc=mu, scale=sigma)

# c. Probabilidade de cauda superior (P(X > 13)) utilizando a Survival Function (sf)
prob_greater_than_13 = norm.sf(13, loc=mu, scale=sigma)

# d. Percentil 90 utilizando a Percent Point Function (ppf)
percentile_90 = norm.ppf(0.90, loc=mu, scale=sigma)

print("=== RESULTADOS DO MODELO NORMAL AJUSTADO ===\n")
print(f"a) P(X <= 10%): {prob_less_than_10:.4f}")
print(f"b) P(10% < X <= 12%): {prob_between_10_and_12:.4f}")
print(f"c) P(X > 13%): {prob_greater_than_13:.4f}")
print(f"d) Percentil 90: {percentile_90:.2f}%")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x = np.linspace(5, 15, 1000) # Intervalo mais relevante para teor alcoólico

# Defina um valor de média (mu) e desvio padrão (sigma) base mais contextuais.
# Usaremos a média e o desvio padrão calculados para o teor alcoólico dos vinhos brancos.
mu_base_alcohol_white = mu_alcohol # Já calculado globalmente
sigma_base_alcohol_white = sigma_alcohol # Já calculado globalmente

# 4. Crie um gráfico para demonstrar o efeito da variação de mu:
plt.figure(figsize=(10, 6))

# Valores para mu mais próximos do contexto real de teor alcoólico
mus_contextual = [mu_base_alcohol_white - 1.5, mu_base_alcohol_white, mu_base_alcohol_white + 1.5]

for mu_val in mus_contextual: # Renomear 'mu' para 'mu_val' para evitar conflito com 'mu' no norm.pdf
    plt.plot(x, norm.pdf(x, loc=mu_val, scale=sigma_base_alcohol_white), label=r'$f(\mu={:.1f}, \sigma={:.1f})$'.format(mu_val, sigma_base_alcohol_white), linewidth=2)

plt.title(r'Efeito da Variação de $\mu$ na Distribuição Normal ($\sigma$ Constante)', fontsize=14)
plt.xlabel('Teor Alcoólico (x)', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# 5. Crie um segundo gráfico para demonstrar o efeito da variação de sigma:
plt.figure(figsize=(10, 6))

# Valores para sigma que mostram diferentes níveis de dispersão, contextualizados
sigmas_contextual = [0.6, sigma_base_alcohol_white, 1.8]

for sigma_val in sigmas_contextual: # Renomear 'sigma' para 'sigma_val' para evitar conflito com 'sigma' no norm.pdf
    plt.plot(x, norm.pdf(x, loc=mu_base_alcohol_white, scale=sigma_val), label=r'$f(\mu={:.1f}, \sigma={:.1f})$'.format(mu_base_alcohol_white, sigma_val), linewidth=2)

plt.title(r'Efeito da Variação de $\sigma$ na Distribuição Normal ($\mu$ Constante)', fontsize=14)
plt.xlabel('Teor Alcoólico (x)', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x_values = np.linspace(5, 15, 1000) # Appropriate range for alcohol content

# Calcule a Função de Densidade de Probabilidade (PDF)
pdf_values = norm.pdf(x_values, loc=mu_alcohol, scale=sigma_alcohol)

# Plot 1: Probabilidade acumulada (P(X <= 10))
plt.figure(figsize=(10, 6))
plt.plot(x_values, pdf_values, color='blue', linewidth=2, label=f'Distribuição Normal (μ={mu_alcohol:.2f}, σ={sigma_alcohol:.2f})')

x_fill_1 = x_values[x_values <= 10]
y_fill_1 = pdf_values[x_values <= 10]
plt.fill_between(x_fill_1, y_fill_1, color='skyblue', alpha=0.5, label=f'P(X <= 10%) = {prob_less_than_10:.4f}')

plt.title('Probabilidade de Álcool Menor ou Igual a 10%', fontsize=16)
plt.xlabel('Teor Alcoólico (%)', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# Plot 2: Probabilidade entre dois valores (P(10 <= X <= 12))
plt.figure(figsize=(10, 6))
plt.plot(x_values, pdf_values, color='blue', linewidth=2, label=f'Distribuição Normal (μ={mu_alcohol:.2f}, σ={sigma_alcohol:.2f})')

x_fill_2 = x_values[(x_values >= 10) & (x_values <= 12)]
y_fill_2 = pdf_values[(x_values >= 10) & (x_values <= 12)]
plt.fill_between(x_fill_2, y_fill_2, color='lightgreen', alpha=0.5, label=f'P(10% <= X <= 12%) = {prob_between_10_and_12:.4f}')

plt.title('Probabilidade de Álcool Entre 10% e 12%', fontsize=16)
plt.xlabel('Teor Alcoólico (%)', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# Plot 3: Probabilidade acima de um valor (P(X > 13))
plt.figure(figsize=(10, 6))
plt.plot(x_values, pdf_values, color='blue', linewidth=2, label=f'Distribuição Normal (μ={mu_alcohol:.2f}, σ={sigma_alcohol:.2f})')

x_fill_3 = x_values[x_values > 13]
y_fill_3 = pdf_values[x_values > 13]
plt.fill_between(x_fill_3, y_fill_3, color='salmon', alpha=0.5, label=f'P(X > 13%) = {prob_greater_than_13:.4f}')

plt.title('Probabilidade de Álcool Maior que 13%', fontsize=16)
plt.xlabel('Teor Alcoólico (%)', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# Plot 4: 90º percentil
plt.figure(figsize=(10, 6))
plt.plot(x_values, pdf_values, color='blue', linewidth=2, label=f'Distribuição Normal (μ={mu_alcohol:.2f}, σ={sigma_alcohol:.2f})')

x_fill_4 = x_values[x_values <= percentile_90]
y_fill_4 = pdf_values[x_values <= percentile_90]
plt.fill_between(x_fill_4, y_fill_4, color='gold', alpha=0.5, label=f'P(X <= {percentile_90:.2f}%) = 0.90')
plt.axvline(percentile_90, color='red', linestyle='--', label=f'90º Percentil = {percentile_90:.2f}%')

plt.title('90º Percentil do Teor Alcoólico', fontsize=16)
plt.xlabel('Teor Alcoólico (%)', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### Condições de Validade e Limitações do Modelo

#### 1. Condições de Validade
* **Dados Contínuos:** O teor alcoólico (%) é uma variável contínua, atendendo ao pressuposto básico de aplicação de distribuições contínuas como a Gaussiana.
* **Comportamento Simétrico:** Espera-se que a distribuição de dados empíricos apresente formato unimodal e aproximadamente simétrico para que a aproximação Normal seja fidedigna.

#### 2. Limitações Estatísticas
* **Fronteiras Físicas vs. Teóricas:** O teor alcoólico possui limites físicos intransponíveis (não pode ser negativo e raramente excede 16% em vinhos de mesa estáveis). A distribuição teórica Normal possui caudas infinitas ($-\infty$ a $+\infty$), gerando probabilidades teóricas não-nulas para valores fisicamente impossíveis.
* **Representatividade Amostral:** Os dados referem-se estritamente aos vinhos brancos "Vinho Verde" de uma região específica de Portugal e produzidos sob condições específicas da safra analisada no estudo de 2009. Não podemos assumir de forma generalizada que essa exata distribuição represente qualquer vinho branco mundial.

#### 3. Avaliação de Aderência Prática
* A adequação do modelo Gaussiano pode ser investigada de forma robusta por meio da visualização do histograma com a curva teórica sobreposta. Testes de hipóteses de normalidade (como Shapiro-Wilk) podem ser complementares, contudo, para amostras grandes ($N = 4898$), eles são extremamente sensíveis a pequeníssimos desvios não-significativos de normalidade, devendo ser interpretados com cautela acadêmica.

*   **Teoria da Distribuição Normal**: A função densidade de probabilidade (PDF) de uma distribuição Normal foi definida como $f(x | \mu, \sigma) = \frac{1}{\sigma \sqrt{2\pi}} e^{ -\frac{1}{2} \left( \frac{x - \mu}{\sigma} \right)^2 }$.
    *   O parâmetro $\mu$ (média) dita o centro da distribuição, deslocando a curva horizontalmente sem alterar sua forma.

    *   **Funções da `scipy.stats.norm`**: `pdf` (densidade de probabilidade), `cdf` (probabilidade acumulada $P(X \leq x)$), `sf` (probabilidade de sobrevivência $P(X > x)$) e `ppf` (função de ponto percentil ou quantil). A função `sf` (*survival function*) calcula a probabilidade de uma variável aleatória assumir valor maior que determinado limite: $P(X>x)$.

## Distribuição t de Student

A distribuição **t de Student** é uma distribuição contínua de probabilidade utilizada principalmente em problemas envolvendo a média de uma população quando o desvio padrão populacional \(\sigma\) é desconhecido e precisa ser estimado a partir da própria amostra.

Sua forma é semelhante à distribuição Normal: é simétrica e centrada em zero. Entretanto, a distribuição t apresenta **caudas mais pesadas**, especialmente quando o número de observações é pequeno. Isso representa a maior incerteza existente quando o desvio padrão populacional é desconhecido.

### Fórmula da Função de Densidade de Probabilidade

A função densidade da distribuição t de Student é:

$$
f(t;\nu)=
\frac{\Gamma\left(\frac{\nu+1}{2}\right)}
{\sqrt{\nu\pi}\Gamma\left(\frac{\nu}{2}\right)}
\left(1+\frac{t^2}{\nu}\right)^{-\frac{\nu+1}{2}}
$$

Onde:

- $$(t)$$ é um possível valor da variável aleatória t;
- $$(\nu)$$ representa os graus de liberdade;
- $$(\Gamma)$$ representa a função gama;
- $$(\pi)$$ representa a constante pi.

### Graus de Liberdade

Quando a distribuição t é aplicada à média de uma única amostra, os graus de liberdade são normalmente calculados por:

$$[
\nu=n-1
]$$

onde $$(n)$$ representa o tamanho da amostra.

Os graus de liberdade determinam o formato da distribuição.

Com poucos graus de liberdade, a curva apresenta caudas mais pesadas. Conforme os graus de liberdade aumentam, a distribuição t se aproxima progressivamente da distribuição Normal padrão.

### Estatística t relacionada à média

Para estudar uma média quando o desvio padrão populacional é desconhecido, pode-se utilizar:

$$[
T =
\frac{\bar{X}-\mu_0}
{S/\sqrt{n}}
]$$

Onde:

- $$(\bar{X})$$ = média da amostra;
- $$(\mu_0)$$ = média populacional considerada como referência;
- $$(S)$$ = desvio padrão da amostra;
- $$(n)$$ = tamanho da amostra.

Neste trabalho, a distribuição t será contextualizada utilizando o **teor alcoólico dos vinhos brancos** da base Wine Quality.

```python
from scipy.stats import t

# Valores possíveis da estatística t
x_t = np.linspace(-5, 5, 1000)

# Diferentes graus de liberdade
graus_liberdade_t = [1, 5, 10, 30]

plt.figure(figsize=(10, 6))

for gl in graus_liberdade_t:
    plt.plot(
        x_t,
        t.pdf(x_t, df=gl),
        linewidth=2,
        label=f't (gl = {gl})'
    )

# Distribuição Normal padrão para comparação
plt.plot(
    x_t,
    norm.pdf(x_t, loc=0, scale=1),
    linestyle='--',
    linewidth=2,
    label='Normal padrão'
)

plt.title(
    'Efeito dos Graus de Liberdade na Distribuição t de Student',
    fontsize=14
)

plt.xlabel('Valor de t', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)

plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```

### Interpretação do Gráfico

O gráfico demonstra que os graus de liberdade influenciam principalmente as caudas da distribuição t.

Quando os graus de liberdade são pequenos, existe maior quantidade de probabilidade nas regiões mais afastadas do centro da distribuição.

À medida que os graus de liberdade aumentam, as caudas tornam-se menos pesadas e a curva aproxima-se da distribuição Normal padrão.

Isso ocorre porque amostras maiores fornecem uma estimativa mais precisa da variabilidade populacional.

Para contextualizar a distribuição t de Student, será utilizada uma amostra de 25 observações do teor alcoólico dos vinhos brancos.

### Pergunta aplicada

**Considerando uma amostra de 25 vinhos brancos, como se comportam as probabilidades associadas à estatística t utilizada em inferências sobre a média do teor alcoólico quando o desvio padrão populacional é desconhecido?**

Nesse contexto, a distribuição t não representa diretamente valores de teor alcoólico. Ela representa uma **estatística padronizada relacionada à média amostral**.

```python
# Seleção de uma amostra reprodutível de 25 vinhos brancos
amostra_t = df_white['alcohol'].sample(
    n=25,
    random_state=42
)

# Estatísticas descritivas da amostra
n_t = len(amostra_t)
media_t = amostra_t.mean()
desvio_t = amostra_t.std(ddof=1)

# Graus de liberdade
gl_t = n_t - 1

print("=== CARACTERÍSTICAS DA AMOSTRA ===\n")

print(f"Tamanho da amostra (n): {n_t}")
print(f"Média amostral: {media_t:.4f}%")
print(f"Desvio padrão amostral: {desvio_t:.4f}%")
print(f"Graus de liberdade: {gl_t}")
```

```python
# a. Probabilidade acumulada P(T <= 1)
prob_t_acumulada = t.cdf(
    1,
    df=gl_t
)

# b. Probabilidade entre dois valores P(-1 < T <= 1)
prob_t_intervalo = (
    t.cdf(1, df=gl_t)
    - t.cdf(-1, df=gl_t)
)

# c. Probabilidade de cauda superior P(T > 2)
prob_t_superior = t.sf(
    2,
    df=gl_t
)

# d. Percentil 95
percentil_t_95 = t.ppf(
    0.95,
    df=gl_t
)

print("=== RESULTADOS DA DISTRIBUIÇÃO t ===\n")

print(f"a) P(T <= 1): {prob_t_acumulada:.4f}")
print(f"b) P(-1 < T <= 1): {prob_t_intervalo:.4f}")
print(f"c) P(T > 2): {prob_t_superior:.4f}")
print(f"d) Percentil 95: {percentil_t_95:.4f}")
```

```python
# Valores da distribuição
x_t_prob = np.linspace(-4, 4, 1000)

pdf_t = t.pdf(
    x_t_prob,
    df=gl_t
)

# GRÁFICO 1 - Probabilidade acumulada

plt.figure(figsize=(10, 6))

plt.plot(
    x_t_prob,
    pdf_t,
    linewidth=2,
    label=f't de Student (gl={gl_t})'
)

mascara = x_t_prob <= 1

plt.fill_between(
    x_t_prob[mascara],
    pdf_t[mascara],
    alpha=0.5,
    label=f'P(T ≤ 1) = {prob_t_acumulada:.4f}'
)

plt.axvline(1, linestyle='--')

plt.title('Probabilidade Acumulada — Distribuição t', fontsize=14)
plt.xlabel('Valor de t', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    x_t_prob,
    pdf_t,
    linewidth=2,
    label=f't de Student (gl={gl_t})'
)

mascara = (x_t_prob >= -1) & (x_t_prob <= 1)

plt.fill_between(
    x_t_prob[mascara],
    pdf_t[mascara],
    alpha=0.5,
    label=f'P(-1 < T ≤ 1) = {prob_t_intervalo:.4f}'
)

plt.axvline(-1, linestyle='--')
plt.axvline(1, linestyle='--')

plt.title('Probabilidade entre -1 e 1 — Distribuição t', fontsize=14)
plt.xlabel('Valor de t', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    x_t_prob,
    pdf_t,
    linewidth=2,
    label=f't de Student (gl={gl_t})'
)

mascara = x_t_prob >= 2

plt.fill_between(
    x_t_prob[mascara],
    pdf_t[mascara],
    alpha=0.5,
    label=f'P(T > 2) = {prob_t_superior:.4f}'
)

plt.axvline(2, linestyle='--')

plt.title('Probabilidade de Cauda Superior — Distribuição t', fontsize=14)
plt.xlabel('Valor de t', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    x_t_prob,
    pdf_t,
    linewidth=2,
    label=f't de Student (gl={gl_t})'
)

mascara = x_t_prob <= percentil_t_95

plt.fill_between(
    x_t_prob[mascara],
    pdf_t[mascara],
    alpha=0.5,
    label='95% da distribuição'
)

plt.axvline(
    percentil_t_95,
    linestyle='--',
    label=f'Percentil 95 = {percentil_t_95:.3f}'
)

plt.title('Percentil 95 — Distribuição t', fontsize=14)
plt.xlabel('Valor de t', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```

A distribuição t é semelhante à Normal, porém apresenta caudas mais pesadas. Essa característica incorpora a incerteza adicional causada pela estimação do desvio padrão populacional. Conforme os graus de liberdade aumentam, a distribuição t aproxima-se da Normal.

**Distribuição Qui-quadrado**

A distribuição Qui-quadrado$$(X^2)$$é uma distribuição contínua definida apenas para valores não negativos.
Diferentemente da distribuição Normal, ela geralmente apresenta assimetria positiva, principalmente quando os graus de liberdade são pequenos.
Sua função densidade é:

$$[
f(x;k)=
\frac{1}{2^{k/2}\Gamma(k/2)}
x^{k/2-1}e^{-x/2}
]$$

para:
$$[
x>0
]$$

Onde:
$$(x)$$ representa um possível valor da estatística Qui-quadrado;

$$(k)$$ representa os graus de liberdade;

$$(\Gamma)$$ representa a função gama.

Quando utilizada para estudar a variância de uma população Normal:

$$[
\chi^2=
\frac{(n-1)S^2}{\sigma^2}
]$$

e:
$$[
k=n-1
]$$

Neste contexto, a distribuição Qui-quadrado não modela diretamente o teor alcoólico. Ela modela uma estatística relacionada à variância dessa característica.

Considerando uma amostra de vinhos brancos, quais probabilidades podem ser associadas à estatística Qui-quadrado utilizada para estudar a variância do teor alcoólico?

**Distribuição F de Fisher.**

A distribuição F de Fisher é uma distribuição contínua utilizada principalmente em procedimentos estatísticos que envolvem razões de variâncias.
Ela possui apenas valores positivos e apresenta assimetria à direita.
Sua função densidade pode ser escrita como:

$$[
f(x;d_1,d_2)=
\frac{
d_2^{d_2/2}d_1^{d_1/2}x^{d_1/2-1}
}{
(d_2+d_1x)^{(d_1+d_2)/2}
B(d_1/2,d_2/2)
}
]$$

para:
$[
x>0
]$

Onde:

$(x)$ representa um possível valor da estatística F;

$(d_1)$ representa os graus de liberdade do numerador;

$(d_2)$ representa os graus de liberdade do denominador;

$(B)$ representa a função beta.

A distribuição F também pode ser compreendida como uma razão entre duas variáveis Qui-quadrado independentes, cada uma dividida por seus respectivos graus de liberdade. NIST
Para comparar as variâncias de dois grupos:

$$[
F=\frac{S_1^2}{S_2^2}
]$$

Para duas amostras:
$$[
df_1=n_1-1
]$$
$$[
df_2=n_2-1
]$$

Portanto, a distribuição F possui dois parâmetros de graus de liberdade, diferentemente da t e da Qui-quadrado.

# **Aplicação real — tinto × branco**

Seleção dos dois grupos

Para a aplicação da distribuição F serão utilizadas observações individuais de dois grupos presentes na base:

Grupo 1: vinhos tintos (df_red).

Grupo 2: vinhos brancos (df_white).

A característica físico-química escolhida será novamente o teor alcoólico (alcohol), permitindo manter coerência com as análises anteriores.

Pergunta aplicada

A variabilidade do teor alcoólico dos vinhos tintos difere da variabilidade observada nos vinhos brancos?
Para investigar essa questão, serão calculadas as variâncias amostrais dos dois grupos e sua razão.

```python
# Observações individuais de teor alcoólico

alcool_red = df_red['alcohol'].dropna()

alcool_white = df_white['alcohol'].dropna()

# Tamanhos das amostras

n_red = len(alcool_red)
n_white = len(alcool_white)

# Variâncias amostrais

var_red = alcool_red.var(ddof=1)
var_white = alcool_white.var(ddof=1)

# Graus de liberdade

df1_f = n_red - 1
df2_f = n_white - 1

# Razão das variâncias

F_observado = var_red / var_white

print("=== COMPARAÇÃO DAS VARIÂNCIAS ===\n")

print(f"Número de vinhos tintos: {n_red}")
print(f"Número de vinhos brancos: {n_white}")

print(f"\nVariância do álcool - tintos: {var_red:.4f}")
print(f"Variância do álcool - brancos: {var_white:.4f}")

print(f"\nF observado: {F_observado:.4f}")

print(f"\ndf1 (numerador): {df1_f}")
print(f"df2 (denominador): {df2_f}")
```

**Interpretação**

Um valor de F igual a 1 representaria variâncias amostrais de mesma magnitude.
Quanto mais a razão se afasta de 1, maior é a diferença relativa entre as variâncias observadas. Isso, isoladamente, não determina significância estatística; a interpretação depende também da distribuição F e de seus graus de liberdade. O teste F clássico utiliza exatamente a razão entre duas variâncias amostrais.

# **Probabilidade acumulada**:

```python
from scipy.stats import f as f_dist

prob_f_acumulada = f_dist.cdf(
    F_observado,
    dfn=df1_f,
    dfd=df2_f
)

print("=== PROBABILIDADE ACUMULADA ===\n")

print(
    f"P(F ≤ {F_observado:.4f}) = "
    f"{prob_f_acumulada:.4f}"
)
```

# **Probabilidade entre dois valores**



```python
from scipy.stats import f as f_dist

limite_f_inferior = 0.8
limite_f_superior = 1.2

prob_f_intervalo = (
    f_dist.cdf(
        limite_f_superior,
        dfn=df1_f,
        dfd=df2_f
    )
    -
    f_dist.cdf(
        limite_f_inferior,
        dfn=df1_f,
        dfd=df2_f
    )
)

print("=== PROBABILIDADE ENTRE DOIS VALORES ===\n")

print(
    f"P({limite_f_inferior} < F ≤ {limite_f_superior}) = "
    f"{prob_f_intervalo:.4f}"
)
```

# **Probabilidade acima de uma valor**

```python
from scipy.stats import f as f_dist

prob_f_superior = f_dist.sf(
    F_observado,
    dfn=df1_f,
    dfd=df2_f
)

print("=== PROBABILIDADE DE CAUDA SUPERIOR ===\n")

print(
    f"P(F > {F_observado:.4f}) = "
    f"{prob_f_superior:.4f}"
)
```

# **Percentil 95**

```python
percentil_f_95 = f_dist.ppf(
    0.95,
    dfn=df1_f,
    dfd=df2_f
)

print("=== PERCENTIL DA DISTRIBUIÇÃO F ===\n")

print(
    f"Percentil 95 = {percentil_f_95:.4f}"
)
```

# **Gráficos sombreados da F**

```python
x_f_prob = np.linspace(
    0.001,
    f_dist.ppf(
        0.999,
        dfn=df1_f,
        dfd=df2_f
    ),
    1500
)

pdf_f = f_dist.pdf(
    x_f_prob,
    dfn=df1_f,
    dfd=df2_f
)

# PROBABILIDADE ACUMULADA

plt.figure(figsize=(10, 6))

plt.plot(
    x_f_prob,
    pdf_f,
    linewidth=2,
    label='Distribuição F'
)

mascara = x_f_prob <= F_observado

plt.fill_between(
    x_f_prob[mascara],
    pdf_f[mascara],
    alpha=0.5,
    label=f'P(F ≤ {F_observado:.3f})'
)

plt.axvline(
    F_observado,
    linestyle='--',
    label=f'F observado = {F_observado:.3f}'
)

plt.title('Probabilidade Acumulada — Distribuição F', fontsize=14)
plt.xlabel('Valor de F', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)

plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```

# **Intervalo**

```python
plt.figure(figsize=(10, 6))

plt.plot(x_f_prob, pdf_f, linewidth=2)

mascara = (
    (x_f_prob >= limite_f_inferior)
    &
    (x_f_prob <= limite_f_superior)
)

plt.fill_between(
    x_f_prob[mascara],
    pdf_f[mascara],
    alpha=0.5,
    label=f'P({limite_f_inferior} < F ≤ {limite_f_superior})'
)

plt.axvline(limite_f_inferior, linestyle='--')
plt.axvline(limite_f_superior, linestyle='--')

plt.title('Probabilidade entre Dois Valores — Distribuição F', fontsize=14)
plt.xlabel('Valor de F')
plt.ylabel('Densidade de Probabilidade')

plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```

# **Cauda superiror**

```python
plt.figure(figsize=(10, 6))

plt.plot(x_f_prob, pdf_f, linewidth=2)

mascara = x_f_prob >= F_observado

plt.fill_between(
    x_f_prob[mascara],
    pdf_f[mascara],
    alpha=0.5,
    label=f'P(F > {F_observado:.3f})'
)

plt.axvline(
    F_observado,
    linestyle='--'
)

plt.title('Probabilidade de Cauda Superior — Distribuição F', fontsize=14)
plt.xlabel('Valor de F')
plt.ylabel('Densidade de Probabilidade')

plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```

# **Percentil**

```python
plt.figure(figsize=(10, 6))

plt.plot(x_f_prob, pdf_f, linewidth=2)

mascara = x_f_prob <= percentil_f_95

plt.fill_between(
    x_f_prob[mascara],
    pdf_f[mascara],
    alpha=0.5,
    label='95% da distribuição'
)

plt.axvline(
    percentil_f_95,
    linestyle='--',
    label=f'Percentil 95 = {percentil_f_95:.3f}'
)

plt.title('Percentil 95 — Distribuição F', fontsize=14)
plt.xlabel('Valor de F')
plt.ylabel('Densidade de Probabilidade')

plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```

# **Condições e limitações da F**

A aplicação clássica do teste F para comparação de duas variâncias pressupõe que as amostras sejam independentes e provenientes de populações aproximadamente Normais.

A Normalidade merece atenção especial, pois o procedimento F clássico para comparação de variâncias é conhecido por ser sensível a desvios dessa condição.

Também é necessário distinguir variância observada na amostra de variância da população. A razão calculada com os dados representa as variâncias amostrais disponíveis e é utilizada para realizar inferências sobre as respectivas variâncias populacionais.

Outro ponto importante é que a distribuição F não representa diretamente valores de teor alcoólico. Ela representa uma estatística construída a partir da razão entre duas estimativas de variância.

# **Comparação entre as Distribuições**

| Distribuição | Quantidade modelada no trabalho | Parâmetros | Formato |
|---|---|---|---|
| Normal | Teor alcoólico dos vinhos brancos | $(\mu)$ e $(\sigma)$ | Simétrica |
| t de Student | Estatística relacionada à média amostral | Graus de liberdade $(\nu$) | Simétrica, com caudas mais pesadas |
| Qui-quadrado | Estatística relacionada à variância | Graus de liberdade $(k)$ | Assimétrica à direita |
| F de Fisher | Razão entre duas variâncias | $(df_1)$ e $(df_2)$| Assimétrica à direita |


**Distribuição Normal**

A Normal foi utilizada para representar diretamente uma característica físico-química: o teor alcoólico dos vinhos brancos.
O parâmetro \(\mu\) controla a posição central da curva, enquanto \(\sigma\) controla sua dispersão.

**Distribuição t de Student**

A distribuição t foi relacionada à média amostral quando o desvio padrão da população não é conhecido.
Os graus de liberdade determinam principalmente o comportamento das caudas. Valores pequenos produzem caudas mais pesadas; valores grandes aproximam a t da Normal.

**Distribuição Qui-quadrado**

A distribuição Qui-quadrado foi relacionada à variância.
Ela não representa diretamente uma medida de teor alcoólico, mas uma estatística construída a partir da variância amostral.
Seus graus de liberdade modificam fortemente a forma e a assimetria da curva.

**Distribuição F de Fisher**

A F foi utilizada para comparar a variabilidade do teor alcoólico entre os vinhos tintos e brancos.
Ela representa uma razão entre duas variâncias e depende simultaneamente dos graus de liberdade do numerador e do denominador.

# **CONCLUSÃO**


A utilização da base Wine Quality permitiu observar que diferentes distribuições de probabilidade respondem a diferentes perguntas estatísticas dentro de um mesmo contexto das Ciências dos Alimentos.

A **distribuição Normal** foi aplicada diretamente ao teor alcoólico dos vinhos brancos. Nesse modelo, a média \(\mu\) determina a localização central da curva, enquanto o desvio padrão \(\sigma\) determina sua dispersão. As áreas calculadas representam probabilidades teóricas de encontrar valores de teor alcoólico em determinadas regiões do modelo ajustado.

A **distribuição t de Student** foi utilizada no contexto da média amostral quando o desvio padrão populacional é desconhecido. Seu principal parâmetro são os graus de liberdade. Com poucos graus de liberdade, a distribuição apresenta caudas mais pesadas; conforme esse número aumenta, a curva aproxima-se da distribuição Normal.

A **distribuição Qui-quadrado** foi aplicada ao contexto da variância. Diferentemente da Normal, ela não modela diretamente uma característica físico-química do vinho, mas uma estatística associada à variabilidade. Os graus de liberdade alteram sua forma, localização e assimetria.

Por fim, a **distribuição F de Fisher** foi utilizada para comparar a variabilidade do teor alcoólico entre vinhos tintos e brancos. A estatística F corresponde a uma razão entre variâncias, e sua distribuição depende dos graus de liberdade do numerador e do denominador.

As áreas sombreadas nos gráficos possuem interpretação probabilística. A área à esquerda de um valor representa uma probabilidade acumulada; a área delimitada por dois valores representa a probabilidade de a estatística encontrar-se naquele intervalo; a área à direita representa uma probabilidade de cauda superior; e um percentil determina o valor abaixo do qual se encontra uma proporção previamente especificada da distribuição.

Dessa forma, a principal diferença entre as quatro distribuições está na **quantidade estatística que cada uma modela**. A Normal foi relacionada aos valores individuais de uma variável físico-química, a t à média, a Qui-quadrado à variância e a F à razão entre duas variâncias.

A análise evidencia que a escolha de uma distribuição de probabilidade deve considerar não apenas o formato dos dados, mas principalmente a pergunta estatística que se pretende responder e as condições necessárias para a aplicação de cada modelo.
